## 1.2 Define a Model with Tunable Parameters

The internal structure of a neural network is composed of individual **neurons**, each of which performs a simple mathematical operation.

The design of these neurons is *loosely inspired* by biological neurons. In the brain, a neuron receives signals through its dendrites, aggregates these inputs, and produces an output (it "fires") based on the combined signal strength.  

In artificial neural networks (ANNs), we mimic this process using mathematical operations, though it’s important to note that the brain does **not** operate in exactly the same way as ANNs.

---

### Structure of a Neuron

A single neuron performs two key operations:

#### 1. Linear Transformation  
Each of the $n$ inputs ($x_i$) is multiplied by a corresponding **weight** ($w_i$), representing how strongly that input influences the neuron’s output.  
These weighted inputs are then summed, and a **bias** term ($b$) is added to shift the result up or down.

This linear step is mathematically equivalent to a multivariate linear regression:

$$
z = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b
$$

---

#### 2. Nonlinear Activation Function  
After the linear combination, the result ($z$) is passed through an **activation function**, which introduces *nonlinearity* into the model.

Without activation functions, a neural network would be equivalent to a single linear model, regardless of how many layers it has.

Two common activation functions are:

- **Sigmoid Function:**  
  $$
  \sigma(z) = \frac{1}{1 + e^{-z}}
  $$  
  The sigmoid function squashes the output into the range $(0, 1)$, making it useful for models that predict probabilities. However, it can cause gradients to vanish in deep networks.

- **ReLU (Rectified Linear Unit):**  
  $$
  \text{ReLU}(z) = \max(0, z)
  $$  
  ReLU is one of the most widely used activation functions today. It introduces nonlinearity while avoiding many of the training difficulties of sigmoid functions, especially in deep networks.

---

In summary, each neuron performs a **linear transformation** followed by a **nonlinear activation**, and by stacking many such neurons into layers, neural networks can model highly complex and nonlinear relationships.

## 1.3 Define a Loss Function

- For **regression tasks** (predicting a continuous scalar value), common choices include **Mean Squared Error (MSE)** or its square root, **Root Mean Squared Error (RMSE)**.

$$
\text{MSE} = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2
$$

## 1.4 Optimize the parameters

Just like with linear regression, we will use **gradient descent** to incrementally decreaese the loss by nudging the model parameters.

With Neural networks, we need to use the **chain rule** to calculate the gradient of each of the parameters. this is called **backpropogation**.

As we increase the complexity of our neural network by combining many neurons together, the chain rule will backpropogate small nudges in the loss function to the corresponding nudges in each of the parameters for each of the neurons.


### Gradient Descent Overview

For each parameter $ \theta $ (e.g., $ w $ or $ b $), gradient descent updates the parameter as:

$$
\theta \leftarrow \theta - \eta \frac{\partial L}{\partial \theta}
$$

where:
- $ \eta $ is the **learning rate** (scalar)
- $ \frac{\partial L}{\partial \theta} $ is the **gradient** of the loss with respect to the parameter.

---

### Backpropagation for a Single Neuron

A single neuron computes:

1. **Linear transformation:**
   $$
   z = w x + b
   $$

2. **Activation (sigmoid):**
   $$
   \hat{y} = \sigma(z) = \frac{1}{1 + e^{-z}}
   $$

3. **Loss (per sample):**
   $$
   L_i = (y_i - \hat{y}_i)^2
   $$

4. **Mean loss (over all samples):**
   $$
   L = \frac{1}{N} \sum_{i=1}^{N} L_i
   $$

---

### Step-by-Step Derivation with the Chain Rule

Let’s now compute the gradient of the loss with respect to the parameters $w$ and $b$.

#### Step 1 — Individual (Per-Sample) Gradient

For a **single sample** $i$, we can use the **chain rule** to compute how the model output depends on each parameter:

$$
\frac{\partial L_i}{\partial w}
=
\frac{\partial L_i}{\partial \hat{y}_i}
\cdot
\frac{\partial \hat{y}_i}{\partial z_i}
\cdot
\frac{\partial z_i}{\partial w}
$$

and

$$
\frac{\partial L_i}{\partial b}
=
\frac{\partial L_i}{\partial \hat{y}_i}
\cdot
\frac{\partial \hat{y}_i}{\partial z_i}
\cdot
\frac{\partial z_i}{\partial b}
$$

Each of these terms is scalar for a single-input neuron.

Breaking it down:

1. **Loss derivative (MSE for one sample):**  
   $$
   \frac{\partial L_i}{\partial \hat{y}_i} = 2(\hat{y}_i - y_i)
   $$

2. **Activation derivative (sigmoid):**  
   $$
   \frac{\partial \hat{y}_i}{\partial z_i} = \hat{y}_i (1 - \hat{y}_i)
   $$

3. **Linear transformation derivatives:**  
   $$
   \frac{\partial z_i}{\partial w} = x_i, \quad \frac{\partial z_i}{\partial b} = 1
   $$

---

#### Step 2 — Mean Gradient Across All Samples

Our **overall loss** $L$ is defined as the **mean** of all $L_i$.  

This comes from the definition of the total loss:

$$
L = \frac{1}{N} \sum_{i=1}^{N} L_i
$$

Taking the derivative:

$$
\frac{\partial L}{\partial w} = \frac{1}{N} \sum_{i=1}^{N} \frac{\partial L_i}{\partial w}, \qquad
\frac{\partial L}{\partial b} = \frac{1}{N} \sum_{i=1}^{N} \frac{\partial L_i}{\partial b}
$$

In other words, the mean of the individual (per-sample) gradients $\frac{\partial L_i}{\partial w}$ yields the full gradient $\frac{\partial L}{\partial w}$.
- Intuitively, this makes sense: every input data point affects the loss function, so to find the full gradient one must consider the ways in wich every single data point nudges the loss function.
- For large datasets, this can be an expensive step, so **stochastic** gradient descent methods will calculate the gradient using only a subset of the full training data.

## 1.4 Stochastic Gradient Descent

Let’s revisit the idea of a **batch** gradient calculation.

- **Intuitively:** every data point nudges the loss; averaging the gradient with respect to each input data point captures the total direction of the gradient.  
  In other words, the **mean of the per-sample gradients** $\frac{\partial L_i}{\partial w}$ yields the full gradient $\frac{\partial L}{\partial w}$:
  $$
  \frac{\partial L}{\partial w} = \frac{1}{N} \sum_{i=1}^{N} \frac{\partial L_i}{\partial w}
  $$

---

### Practical Motivation for Stochastic Methods

While this formulation is mathematically precise, it quickly becomes **computationally expensive** as datasets grow.

#### 1. Computational cost
To perform one gradient descent update in **batch gradient descent**, we must:
1. Compute model predictions for **every sample** in the training set.  
2. Compute the gradient of the loss with respect to each parameter for **every sample**.  
3. Take the mean across all $N$ samples to obtain the gradient direction.  

When $N$ is large (e.g., millions of data points), a single gradient update requires passing the **entire dataset** through the model.  

This makes learning extremely **slow and memory-intensive**, especially when using deep neural networks with millions of parameters.

---

#### 2. Delayed learning signal
Because batch gradient descent waits until it has processed the *entire* dataset before updating parameters.

This means the model must complete a full pass over the data before making any progress toward updating the weights and biases and minimizing the loss — a huge bottleneck for large-scale or streaming data.

---

#### 3. Stochastic gradient descent as an approximation
Instead of computing the *exact* full gradient, we can use an **approximation** based on a *subset* of the data.

- **Stochastic Gradient Descent (SGD)** uses **one random sample** at a time:
  $$
  \frac{\partial L}{\partial w} \approx \frac{\partial L_i}{\partial w}
  $$
  Each update is fast, cheap, and provides a *noisy* but often sufficient estimate of the true gradient direction.

- **Mini-Batch Gradient Descent** generalizes this idea by computing the gradient over a **small subset (batch)** of the data:
  $$
  \frac{\partial L}{\partial w} \approx \frac{1}{B} \sum_{i \in \mathcal{B}} \frac{\partial L_i}{\partial w}
  $$
  where $\mathcal{B}$ is a randomly chosen batch of size $B \ll N$.  
  This approach balances the **efficiency of SGD** with the **stability of batch gradient descent**.

---

#### 4. Key trade-offs
| Method | Data Used per Update | Computational Cost per Step | Update Frequency | Noise Level | Typical Use |
|---------|----------------------|-----------------------------|------------------|--------------|--------------|
| **Batch GD** | All $N$ samples | Very high | Low (1 per epoch) | None | Small datasets |
| **SGD** | 1 sample | Very low | Very high | High | Online / large-scale data |
| **Mini-Batch GD** | $B$ samples | Moderate | Moderate | Moderate | Deep learning default |

---

#### 5. Why stochasticity helps
Interestingly, the **noise** introduced by stochastic updates can actually be beneficial:
- It allows the optimizer to **explore the loss landscape** instead of getting stuck in shallow local minima.  
- The randomness acts like a “thermal” effect, helping parameters jump out of plateaus and saddle points.  
- In practice, this often leads to **faster convergence** to a good solution — even if it’s not the exact global minimum.

---

#### What Is an Epoch?

An **epoch** refers to **one complete pass** through the entire training dataset.

During an epoch, the optimizer has the opportunity to see *every training example once* and update the model parameters accordingly.  
How many updates occur *within* an epoch depends on the batching strategy:

- **Batch Gradient Descent:**  
  The entire dataset is processed as one batch → **1 update per epoch**.

- **Mini-Batch Gradient Descent:**  
  The dataset is divided into smaller batches of size $B$.  
  If there are $N$ total samples, then each epoch consists of $\frac{N}{B}$ updates.  
  For example, with $N=10{,}000$ samples and $B=100$, there are **100 updates per epoch**.

- **Stochastic Gradient Descent:**  
  Each sample forms its own batch ($B=1$), so there are **$N$ updates per epoch**.

---

#### Why Train for Multiple Epochs?

Training typically requires **many epochs**, because one pass through the data is rarely enough for the model to converge.

- Early epochs usually cause large improvements as parameters move toward better regions of the loss surface.  
- Later epochs make smaller refinements, fine-tuning the model around a local minimum.  
- After too many epochs, the model may begin to **overfit**, memorizing noise in the training set.

---

### Summary

- Computing the exact gradient across all data points is **expensive** and **slow**.  
- **Stochastic** and **mini-batch** gradient descent approximate the true gradient using fewer samples, dramatically improving efficiency.  
- The resulting “noisy” gradients still point *on average* in the right direction, enabling faster and more scalable learning.  
- This trade-off — between precision and speed — is at the heart of modern deep learning optimization.

PyTorch tensors might look like NumPy arrays, but they come with two powerful superpowers that make them essential for deep learning:

1. **Automatic Differentiation**
Tensors can keep track of every operation applied to them.

- set `requires_grad=True` when creating a tensor, and PyTorch builds a computational graph behind the scenes
- call `.backward()` on the tensor and it automatically computes all the gradients you need for training.

**No more hand-writing derivative formulas using the chain rule!**

Let's see this in action:

````python
import torch

# Create a tensor that tracks gradients
x = torch.tensor(2.0, requires_grad=True)
y = x**2 + 3*x + 1  # some arbitrary function

# Compute gradient (dy/dx)
y.backward()

print("Value of y:", y.item())
print("Gradient dy/dx:", x.grad.item())
````

Did autograd give the correct answer for the derivative? We can check, in this simple case, by manual differentiation and plugging in the value of `x`:

y = x^2 + 3x +1
$$
$$
\frac{dy}{dx} = 2x +3
$$
$$
\frac{dy}{dx} @ (x=2) = 2*2 +3=7
$$

## 2.4 Optimizers and Training

With automatic gradients from PyTorch, training using gradient descent can be done in only a few lines of code:
1. Compute a loss
2. Call loss.backward() to populate parameter gradients,
3. Define a pytorch optimizer and call `optimizer.step()` to update the model parameters

```python
# a training step in three lines of python code:
optimizer.zero_grad()  # 1) clear old grads
loss.backward()        # 2) compute new grads via autograd
optimizer.step()       # 3) update params
```

Why zero_grad? Remember, gradients accumulate by default so you can average over mini-batches. If you don’t clear them, you’ll unintentionally sum across steps.

---

### What optimizers do

At each update step, an optimizer:

- Reads gradients from each parameter’s .grad (populated by loss.backward()).

- Applies an update rule (e.g. gradient descent) to modify parameters.

- Optionally regularizes (e.g., weight decay / L2).

PyTorch offers several different optimizers beyond simple gradient descent. These optimizers can lead to faster training. Two examples include:

- **Stochastic gradient descent:** `torch.optim.SGD(params, lr, momentum=0.0, nesterov=False, weight_decay=0.0)`

- **Adam:** `torch.optim.Adam(params, lr=1e-3, betas=(0.9, 0.999), weight_decay=0.0)` Adam provides adaptive per-parameter step sizes; it is an excellent default for many problems.

---

### `model.train()` vs `model.eval()`

In PyTorch, your model can run in two different modes — training mode and evaluation mode.
You can switch between them with:

```python
model.train()   # training mode
model.eval()    # evaluation (testing) mode
```

These modes don’t change the model’s parameters, but they tell PyTorch how the model should behave depending on what you’re doing.

### `model.train()`

Use this mode when training your model — that is, when you’re looping through the data, computing the loss, calling loss.backward(), and updating the parameters with the optimizer.

In this mode, the model is allowed to behave in ways that help it learn, such as using randomness or keeping track of information about the training data.

### `model.eval()`

Use this mode when testing or evaluating your model — for example, when you want to check how well it performs on unseen data.
In evaluation mode, the model behaves in a stable and consistent way: it turns off any randomness used during training so that predictions are reliable and repeatable.

## 3.2 Model Width and the Universal Approximation Theorem

### What is Model Width?

**Model width** refers to the number of neurons in a layer of a neural network. In our architecture with a single hidden layer, the width is simply the number of hidden neurons (`n`).

- **Narrow networks**: Few neurons (e.g., n=2 or n=5)
- **Wide networks**: Many neurons (e.g., n=50 or n=100)

The width of a network directly affects its **capacity** - its ability to learn complex patterns and relationships in data.

### The Universal Approximation Theorem

One of the most remarkable results in neural network theory is the **Universal Approximation Theorem**, which states:

> A feedforward neural network with a single hidden layer containing a finite number of neurons can approximate any continuous function on a compact subset of ℝⁿ to arbitrary precision, given appropriate activation functions.

**What does this mean in practice?**

1. **Single hidden layer is enough (theoretically)**:
You can model most well-behaved functions with only a single hidden layer, as long as it contains enough neurons.

2. **Width provides expressiveness**: By increasing the number of hidden neurons, you can approximate increasingly complex functions with arbitrary accuracy.

3. **The catch**: While the theorem guarantees that such an approximation *exists*, it doesn't tell us:
   - How many neurons we need (could be impractically large)
   - How to find the right weights (training may be difficult)
   - Whether the approximation generalizes well to new data

---

### Practical Implications

In practice, this means:
- **Start simple**: Begin with a small number of neurons and increase if needed
- **Width vs. Depth trade-off**: While wide shallow networks can theoretically approximate any function, deep networks (multiple hidden layers) often achieve better results with fewer total parameters
- **Experimentation is key**: The "right" width depends on your specific problem and dataset

## 3.3 Model Regularization

### Why not just add more neurons?

The Universal Approximation Theorem tells us that a sufficiently wide single hidden layer can approximate any continuous function. This might lead to a natural question:

> **"Why not just use 100,000 hidden neurons for every problem?"**

The answer lies in one of the central challenges of machine learning: **overfitting**.

---

### Overfitting: When Models Memorize Instead of Learn

When we make our models very wide and/or very deep (many neurons), they gain tremendous **expressive capacity** - the ability to fit increasingly complex patterns. However, this becomes a double-edged sword:

- ✓ **Good:** The model can capture subtle, genuine patterns in the data
- ✗ **Bad:** The model can also memorize noise and random fluctuations specific to the training data

A model that overfits achieves:
- **Low training loss** (fits training data perfectly)
- **High test loss** (fails on new, unseen data)

This is like a student who memorizes answers to practice problems but doesn't understand the underlying concepts - they fail when faced with new questions.

### Visualizing the Problem

Imagine fitting our Gaussian bumps dataset:
- **Underfit model** (too few neurons): Misses the bumps entirely, smooth but inaccurate
- **Well-fit model** (just right): Captures the bumps, generalizes to new data
- **Overfit model** (too many neurons): Follows every noise wiggle in training data, wild predictions elsewhere

The **train-test gap** - the difference between training loss and test loss - is our warning sign for overfitting.

---

## Regularization: Teaching Models to Generalize

**Regularization** techniques help prevent overfitting by constraining the model during training. Think of it as adding "guardrails" that prevent the model from becoming too complex or too confident in spurious patterns.

We'll explore two powerful regularization techniques:

### 1. Dropout ⬇

**What it does:** During training, randomly "drops out" (sets to zero) a fraction of neurons in each training step.

**How it works:**
```
Without dropout:          With dropout (50%):
All neurons active        Random neurons disabled

[●] [●] [●] [●]          [●] [○] [●] [○]
 │   │   │   │            │       │
 └───┴───┴───┘            └───────┘
```

**Why it helps:**
- Prevents neurons from co-adapting (relying too heavily on each other)
- Forces the network to learn robust features that work even when some neurons are missing
- Acts like training an ensemble of smaller networks
- At test time, all neurons are active (no dropout), but their outputs are scaled appropriately

**In our code:**
- `dropout_rate=0.0`: No dropout (default)
- `dropout_rate=0.3`: 30% of neurons randomly dropped during training
- `dropout_rate=0.5`: 50% of neurons randomly dropped during training

**When to use:** Especially effective with wide networks (many neurons) that show signs of overfitting.

### 2. Weight Decay (L2 Regularization) ⚖️

**What it does:** Penalizes large weight values by adding a term to the loss function.

**How it works:**
```
Regular loss:              Loss with weight decay:
L = MSE(y, ŷ)             L = MSE(y, ŷ) + λ·Σ(w²)
                                         ↑
                                   penalty for large weights
```

Where:
- `λ` (lambda) is the weight decay parameter (strength of regularization)
- `Σ(w²)` is the sum of squared weights

**Why it helps:**
- Encourages smaller, more distributed weight values
- Prevents any single feature or neuron from dominating
- Leads to simpler, smoother decision boundaries
- Implements "Occam's Razor" - prefer simpler explanations

**In our code:**
- `weight_decay=0.0`: No weight decay (default)
- `weight_decay=0.001`: Mild regularization
- `weight_decay=0.01`: Strong regularization

**When to use:** Helpful when you notice large fluctuations in the model's predictions or a significant train-test gap.
